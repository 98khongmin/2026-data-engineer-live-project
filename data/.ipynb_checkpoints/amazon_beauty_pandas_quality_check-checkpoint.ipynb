## 1. 데이터 경로 및 테스트 옵션

`DATA_PATH`에는 Hugging Face에서 받은 리뷰 JSONL 파일 경로를 입력하세요. `SAMPLE_ROWS=None`이면 전체 로드(약 70만 건), 정수를 주면 그 행 수만 불러옵니다. 표본 검사는 전체 중복 건수를 확정할 수 없습니다.

```python
DATA_PATH = Path("../data/All_Beauty.jsonl")  # 필요에 따라 .jsonl.gz도 가능
SAMPLE_ROWS = None  # 빠른 확인: 10000 / 전체 검사: None
EXPORT_RESULTS = False  # True면 현재 폴더의 quality_output/에 CSV 저장
OUTPUT_DIR = Path("quality_output")

# 이 범위는 사용자 보고서의 분석 대상 범위입니다. 실제 원본 min/max도 별도 출력합니다.
DATE_START = pd.Timestamp("2000-11-01", tz="UTC")
DATE_END_EXCLUSIVE = pd.Timestamp("2023-10-01", tz="UTC")
```

## 2. 원본 불러오기

실제 파일이 없을 경우 *정상 데이터, 완전 중복, 키 충돌, 빈 사용자 ID, 날짜 오류*를 포함한 소형 예제를 생성합니다.

```python
def load_jsonl(path: Path, nrows=None):
    opener = gzip.open if str(path).endswith(".gz") else open
    records = []
    with opener(path, "rt", encoding="utf-8") as f:
        for idx, line in enumerate(f):
            if nrows is not None and idx >= nrows:
                break
            if line.strip():
                records.append(json.loads(line))
    return pd.DataFrame.from_records(records)

if DATA_PATH.exists():
    df = load_jsonl(DATA_PATH, SAMPLE_ROWS)
    source_label = f"실제 파일: {DATA_PATH}"
else:
    df = None
    source_label = "내장 가상 테스트 데이터 (실제 결과 아님)"
print(source_label, "| rows:", len(df))
display(df.head())
```

## 3. `user_id` / 상품 ID / `timestamp` 정합성 검증

- `user_id`, `parent_asin`: NULL/공백만 있는 문자열은 불량.
- `timestamp`: 숫자로 파싱 가능한 정수 밀리초, 2000-11-01 00:00 UTC 이상 ~ 2023-10-01 00:00 UTC 미만.
- 임의의 문자열 값을 비정상적으로 '수정'하지 않습니다. 실패 행은 별도로 남깁니다.
- 날짜 기준은 과제에서 사용하는 분석 범위이며 원본 파일 실제 범위와 다르면 조정해야 합니다.

```python
# ==========================================
# 1. 데이터에 필수 컬럼이 존재하는지 확인
# ==========================================

# 리뷰를 구분하는 데 필요한 3개의 컬럼
required = ["user_id", "parent_asin", "timestamp"]

# 필요한 컬럼 중 데이터프레임(df)에 없는 컬럼을 찾음
# 예: timestamp가 없다면 missing_cols = ["timestamp"]
missing_cols = [c for c in required if c not in df.columns]

# 필수 컬럼이 하나라도 없으면 오류를 발생시키고 실행 중단
if missing_cols:
    raise ValueError(f"원본에 필수 컬럼이 없습니다: {missing_cols}")


# ==========================================
# 2. 원본 데이터 복사
# ==========================================

# 원본 df를 변경하지 않도록 복사본 생성
# 이후 데이터 검사는 work에서 진행
work = df.copy()


# ==========================================
# 3. user_id와 parent_asin 유효성 검사
# ==========================================

# user_id와 parent_asin을 각각 검사
for c in ["user_id", "parent_asin"]:

    # 문자열로 변환한 뒤 앞뒤 공백 제거
    # 예: " ABC123 " → "ABC123"
    # 새 컬럼: user_id_clean, parent_asin_clean
    work[c + "_clean"] = work[c].astype("string").str.strip()

    # 값이 비어 있거나 누락되었는지 검사
    # notna() : NULL이 아니면 True
    # ne("")  : 빈 문자열이 아니면 True
    # fillna(False) : 검사 중 발생한 결측 결과를 False로 처리
    #
    # 두 조건을 모두 만족하면 True(정상)
    # 하나라도 만족하지 않으면 False(비정상)
    #
    # 새 컬럼: user_id_valid, parent_asin_valid
    work[c + "_valid"] = (
        work[c + "_clean"].notna()
        & work[c + "_clean"].ne("").fillna(False)
    )


# ==========================================
# 4. timestamp 숫자 형식 검사
# ==========================================

# timestamp를 숫자로 변환
# 숫자로 변환할 수 없는 값은 NaN(결측값) 처리
# 예: "1588687728923" → 1588687728923
# 예: "abc" → NaN
raw_ts = pd.to_numeric(work["timestamp"], errors="coerce")

# timestamp가 정상적인 정수인지 검사
#
# raw_ts.notna() : 숫자로 변환 가능한 값인지
# raw_ts >= 0    : 0 이상의 숫자인지
# raw_ts % 1 == 0 : 소수점이 없는 정수인지
#
# 예:
# 1588687728923 → True
# -100           → False
# 123.5          → False
# "abc"          → False
work["timestamp_valid_integer"] = (
    raw_ts.notna()
    & (raw_ts >= 0)
    & (raw_ts % 1 == 0)
)


# ==========================================
# 5. timestamp를 날짜로 변환
# ==========================================

# Amazon 리뷰의 timestamp는 밀리초(ms) 단위
#
# unit="ms"  : 밀리초 단위로 해석
# utc=True   : UTC 시간 기준
# errors="coerce" : 변환할 수 없는 값은 NaT(날짜 결측값) 처리
#
# 예:
# 1588687728923 → 2020-05-05 15:28:48.923 UTC
work["review_datetime_utc"] = pd.to_datetime(
    raw_ts,
    unit="ms",
    utc=True,
    errors="coerce"
)


# ==========================================
# 6. timestamp 날짜 범위 검사
# ==========================================

# DATE_START와 DATE_END_EXCLUSIVE는
# 노트북 앞부분에서 미리 설정한 분석 기간
#
# 예:
# DATE_START = pd.Timestamp("2000-11-01", tz="UTC")
# DATE_END_EXCLUSIVE = pd.Timestamp("2023-10-01", tz="UTC")

# ge() : 이상 (>=)
# lt() : 미만 (<)
#
# 2000년 11월 1일 이상이고
# 2023년 10월 1일 미만인 날짜만 정상으로 판단
#
# 2023년 9월 30일까지 포함하려는 목적
work["timestamp_valid_range"] = (
    work["review_datetime_utc"].ge(DATE_START)
    & work["review_datetime_utc"].lt(DATE_END_EXCLUSIVE)
)


# ==========================================
# 7. timestamp 최종 유효성 검사
# ==========================================

# timestamp가 정수 형식이고
# 날짜 범위에도 포함되어야 정상
#
# 두 조건 모두 True → True
# 하나라도 False → False
work["timestamp_valid"] = (
    work["timestamp_valid_integer"]
    & work["timestamp_valid_range"]
)


# ==========================================
# 8. 리뷰 식별 키 전체 검증
# ==========================================

# 리뷰 식별에 필요한 3개 항목을 모두 검사
#
# 1. user_id가 정상인가?
# 2. parent_asin이 정상인가?
# 3. timestamp가 정상인가?
#
# 3개가 모두 True여야 valid_key = True
#
# 즉, 리뷰 고유 ID를 생성할 수 있는 데이터인지 판단
work["valid_key"] = (
    work["user_id_valid"]
    & work["parent_asin_valid"]
    & work["timestamp_valid"]
)


# ==========================================
# 9. 검증 결과 요약표 생성
# ==========================================

summary = pd.DataFrame({

    # 결과표에 표시될 검사 항목
    "검증 항목": [
        "전체 리뷰",
        "user_id 누락/빈값",
        "parent_asin 누락/빈값",
        "timestamp 정수/양수 오류",
        "timestamp 기간 오류",
        "식별 키 정상",
        "식별 키 비정상"
    ],

    # 각 검사 항목에 해당하는 리뷰 개수 계산
    "건수": [

        len(work),  # 전체 리뷰 개수

        # user_id가 비정상인 리뷰 개수
        int((~work["user_id_valid"]).sum()),

        # parent_asin이 비정상인 리뷰 개수
        int((~work["parent_asin_valid"]).sum()),

        # timestamp가 정수 또는 0 이상 조건을 위반한 개수
        int((~work["timestamp_valid_integer"]).sum()),

        # timestamp가 정수 조건은 만족하지만
        # 허용 날짜 범위를 벗어난 리뷰 개수
        int((
            work["timestamp_valid_integer"]
            & ~work["timestamp_valid_range"]
        ).sum()),

        # 식별 키 3개가 모두 정상인 리뷰 개수
        int(work["valid_key"].sum()),

        # 식별 키 중 하나 이상이 비정상인 리뷰 개수
        int((~work["valid_key"]).sum())
    ]
})


# ==========================================
# 10. 결과 출력
# ==========================================

# 검증 결과를 표로 출력
display(summary)

# 리뷰 날짜 중 가장 오래된 날짜 출력
print("실제 유효 날짜 최솟값:", work["review_datetime_utc"].min())

# 리뷰 날짜 중 가장 최근 날짜 출력
print("실제 유효 날짜 최댓값:", work["review_datetime_utc"].max())
```

## 4. 해시 기반 `review_id` 생성

유효 키에서 `(user_id, parent_asin, timestamp)`를 **길이 기반 직렬화** 후 SHA-256으로 해시합니다. 단순 문자열 구분자의 우발적 충돌을 피하기 위해 JSON 배열로 직렬화합니다. 같은 세 값이면 같은 해시가 나오며, 해시값이 같아도 리뷰 본문이 서로 다를 수 있으므로 충돌 검사를 반드시 수행해야 합니다.

```python
import uuid
import json
import pandas as pd

# 1. UUID 생성 함수
def make_review_id(user_id, parent_asin, timestamp):

    # 리뷰 식별에 필요한 3개 값을 하나로 묶음
    payload = json.dumps(
        [str(user_id), str(parent_asin), int(timestamp)],
        ensure_ascii=False,
        separators=(",", ":")
    )

    # 같은 입력값이면 항상 같은 UUID 생성
    return str(uuid.uuid5(uuid.NAMESPACE_URL, payload))


# 2. 유효한 리뷰만 추출
valid = work.loc[work["valid_key"]].copy()


# 3. timestamp를 정수형으로 변환
valid["timestamp_ms"] = (
    pd.to_numeric(valid["timestamp"]).astype("int64")
)


# 4. 각 리뷰에 UUID 생성
valid["review_id"] = [
    make_review_id(u, p, t)
    for u, p, t in zip(
        valid["user_id_clean"],
        valid["parent_asin_clean"],
        valid["timestamp_ms"]
    )
]


# 5. 전체 리뷰 수와 고유 UUID 개수 비교
print("유효 리뷰:", len(valid))
print("고유 review_id:", valid["review_id"].nunique())


# 6. 생성된 UUID 확인
display(
    valid[
        ["user_id", "parent_asin", "timestamp", "review_id"]
    ].head()
)
```

## 8. 보고서에 적을 때

숫자는 반드시 **실제 파일 전체를 사용해 실행한 결과**로 교체하세요.

- **식별 규칙**: `SHA256(JSON([user_id, parent_asin, timestamp_ms]))`
- **정합성 기준**: 빈 `user_id`/`parent_asin` 제외, timestamp 정수 밀리초·UTC 기간 검증
- **중복 정책**: 동일 키+동일 내용만 제거, 동일 키+상이 내용은 보류/별도 조사
- **재처리 대응**: 운영 PostgreSQL의 `review_id UNIQUE` + UPSERT 사용. *단, 충돌을 검증하지 않은 채 UPSERT로 덮어쓰면 안 됨.*

**주의:** 본 규칙은 분석 목적에 맞춘 후보 키이며, 원본에서 리뷰 고유성을 공식적으로 보증하는 식별자는 아닙니다. 데이터의 `user_id`만으로는 중복 판단을 할 수 없습니다.
